```python
import numpy as np
import time
```

```python
#performance
size=10_000_000

#python list
py_list=(range(size))
start=time.time()
sq_list=[i*2 for i in py_list]
end=time.time()
print(f"python list time ={end-start} seconds")

#numpy arrys
np_array=np.array(py_list)
start=time.time()
sq_array=np_array*2
end=time.time()
print(f"numpy array time ={end-start} seconds")
```

```python
#memory
import sys

print(f"python list size ={sys.getsizeof(py_list)*len(py_list)}")
print(f"numpy array size ={np_array.nbytes}") 
```
