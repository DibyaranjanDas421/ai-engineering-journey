```python
import numpy as np

##Common NumPy data types

| Data type | Meaning | Example |
|---|---|---|
| `int8`, `int16`, `int32`, `int64` | Signed integers of different sizes | `10`, `-25` |
| `uint8`, `uint16`, `uint32`, `uint64` | Unsigned integers (non-negative) | `0`, `255` |
| `float16`, `float32`, `float64` | Floating-point numbers | `3.14`, `2.5` |
| `complex64`, `complex128` | Complex numbers | `2 + 3j` |
| `bool` | Boolean values | `True`, `False` |
| String dtype | Text values | `"Python"` |
| `object` | Python objects | Mixed Python objects |

##Why do data types matter in AI and Machine Learning?

- **Memory efficiency:** `float32` uses half the element storage of `float64`.
- **Performance:** Choosing a suitable dtype can improve computational efficiency.
- **Model compatibility:** Many machine learning and deep learning workflows use `float32` for numerical inputs and model parameters.
- **Data correctness:** Choosing the right type helps avoid unwanted conversions and precision loss.

For example, `float32` is commonly used for model inputs and weights, while integer types such as `int32` and `int64` are often used for labels and indices.
```

```python
#int64
arr1=np.array([1,2,3,4])
print(arr1,arr1.dtype)
```

```python
#float64
arr2=np.array([1,2,3,4.0])
print(arr2,arr2.dtype)
```

```python
#U32
arr3=np.array([1,2,3,4.0,"Dibya"])
print(arr3,arr3.dtype)
```

```python
#complex128
arr4=np.array([2+3j])
arr5=np.array([3+2j])
print(arr4+arr5,arr4.dtype)
```

```python
#object
arr6=np.array([1,2,"Dibya",{"city":"Bhadrak"}])
print(arr6,arr6.dtype)
```
