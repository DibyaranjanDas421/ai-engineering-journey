````python
# NumPy Broadcasting

## 1. What is Broadcasting?

**Broadcasting** is a NumPy feature that allows arithmetic operations between arrays of different shapes, provided their dimensions are compatible.

If two arrays have incompatible shapes, element-wise operations such as addition, subtraction, multiplication, and division generally raise a `ValueError`.

**Important:** Broadcasting is not limited to 1D arrays. It also works with compatible 2D and 3D arrays.

## 2. Rules of Broadcasting

NumPy compares the shapes of two arrays from **right to left**, starting with the last dimension.

Two dimensions are compatible if:

1. Both dimensions are equal, OR
2. One of the dimensions is `1`.

If one array has fewer dimensions, NumPy treats the missing dimensions on the left as `1` for comparison.

If these conditions are not satisfied, broadcasting fails.

## 3. Example: Broadcasting a 1D Array with a 2D Array

### Code

Run this in a **Code cell**:

```python
import numpy as np

arr2D = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

arr1D = np.array([1, 2, 3])

result = arr2D + arr1D

print(result)
print("Shape:", result.shape)
```

### Output

```text
[[11 22 33]
 [41 52 63]]
Shape: (2, 3)
```

### Explanation

- `arr2D.shape` = `(2, 3)`
- `arr1D.shape` = `(3,)`
- For broadcasting, NumPy treats the 1D array's shape as `(1, 3)` for comparison.

Compare the dimensions from right to left:

| Dimension | 2D array | 1D array (adjusted) | Compatible? |
|---|---:|---:|---|
| Last | 3 | 3 | Yes, equal |
| First | 2 | 1 | Yes, one is 1 |

Therefore, broadcasting works, and the result has shape `(2, 3)`.

Conceptually, NumPy adds the 1D array to each row:

```text
[[10, 20, 30],       [[1, 2, 3],
 [40, 50, 60]]   +    [1, 2, 3]]
```

NumPy can perform this without creating actual repeated copies of the 1D array.

## 4. Example: Incompatible Arrays

### Code

Run this in a separate **Code cell**:

```python
arr1 = np.array([1, 2, 3])   # Shape: (3,)
arr2 = np.array([10, 20])    # Shape: (2,)

result = arr1 + arr2
```

### Output

```text
ValueError: operands could not be broadcast together with shapes (3,) (2,)
```

### Explanation

- First array shape: `(3,)`
- Second array shape: `(2,)`
- The dimensions `3` and `2` are unequal, and neither is `1`.

Therefore, broadcasting is not possible.

## 5. Quick Examples

| Shape A | Shape B | Broadcasting | Result Shape |
|---|---|---|---|
| `(2, 3)` | `(3,)` | Possible | `(2, 3)` |
| `(2, 3)` | `(2, 1)` | Possible | `(2, 3)` |
| `(3, 1)` | `(1, 4)` | Possible | `(3, 4)` |
| `(2, 3)` | `(2,)` | Not possible | — |
| `(3,)` | `(2,)` | Not possible | — |

## 6. Quick Revision

- Broadcasting allows element-wise operations between arrays of compatible shapes.
- Compare dimensions from right to left.
- Dimensions must be equal or one of them must be `1`.
- Broadcasting is not restricted to 1D arrays.
- Incompatible shapes generally cause a `ValueError`.
- **Array creation and broadcasting are different:** nested lists with unequal row lengths can raise a `ValueError` while creating a regular NumPy array, before broadcasting is even attempted.
````
