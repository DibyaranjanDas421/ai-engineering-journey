```python
import numpy as np
```

```python
arr=np.array([[1,2],[3,4]])

mean=np.mean(arr)
std_dev=np.std(arr)
normalized_arr=(arr-mean)/std_dev
print(normalized_arr)
print(mean)
print(std_dev)
```

```python
# ==================================================
# MEAN, STANDARD DEVIATION AND STANDARDIZATION
# ==================================================

# 1. MEAN (AVERAGE)
# -----------------
# Mean is the sum of all values divided by the total number of values.
#
# Formula:
# Mean = Sum of all values / Total number of values
#
# Mathematical formula:
# μ = Σx / N
#
# Example:
# arr = [1, 2, 3, 4]
# Mean = (1 + 2 + 3 + 4) / 4 = 2.5
#
# NumPy:
# mean = np.mean(arr)


# 2. STANDARD DEVIATION
# ---------------------
# Standard deviation measures how far the values are spread out
# from their mean.
#
# Formula:
# σ = sqrt(Σ(x - μ)² / N)
#
# Where:
# x  = Each value
# μ  = Mean
# N  = Total number of values
# σ  = Population standard deviation
#
# NumPy:
# std_dev = np.std(arr)
#
# Note:
# np.std() calculates population standard deviation by default.


# 3. STANDARDIZATION (Z-SCORE)
# ----------------------------
# Standardization transforms data so that:
# Mean becomes approximately 0
# Standard deviation becomes 1
#
# Formula:
# z = (x - μ) / σ
#
# Where:
# x = Original value
# μ = Mean
# σ = Standard deviation
# z = Standardized value
#
# NumPy:
# normalized_arr = (arr - mean) / std_dev
#
# IMPORTANT:
# Use mean, not min.
# min is a Python built-in function.
#
# Standardization is undefined when the standard deviation is zero.
# ==================================================
```
